```python
#find common elements between two lists using set 
# Find common elements between two lists using sets

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

common = set(list1).intersection(set(list2))

print("Common elements:", common)
```

```python
#3 courses have following students
#find the students who are enrolled in 2 courses
python = {'a', 'b', 'c', 'd', 'e'}
java = {'c', 'd', 'f', 'g'}
cpp = {'d', 'e', 'h', 'i'}

two_courses = (
    python.intersection(java)
    .union(python.intersection(cpp))
    .union(java.intersection(cpp))
) - python.intersection(java).intersection(cpp)

print(two_courses)
```

```python
#two documents contain:

'''doc1 = "python programming is easy to learn"
doc = python programming is easy and interesting'''

doc1 = "python programming is easy to learn"
doc2 = "python programming is easy and interesting"

set1 = set(doc1.split())
set2 = set(doc2.split())

common_words = set1.intersection(set2)

unique_words = set1.union(set2)

similarity = (len(common_words) / len(unique_words)) * 100

print("Common words:", common_words)
print("Total unique words:", len(unique_words))
print("Similarity:", similarity, "%")

#calculate the similarity % using :
#similarity = (common words/total unique words) * 100
```
