```python
Q1. Count vowels, consonants, digits and special characters 

Write a program to analyze:

let str = "Hello JavaScript 2026!";
 
 
```

```python
let str = "Hello JavaScript 2026!";

let vowels = 0;
let consonants = 0;
let digits = 0;
let special = 0;

for (let char of str) {
    if ("aeiouAEIOU".includes(char)) {
        vowels++;
    } 
    else if (/[a-zA-Z]/.test(char)) {
        consonants++;
    } 
    else if (/[0-9]/.test(char)) {
        digits++;
    } 
    else if (char !== " ") {
        special++;
    }
}

console.log("Vowels:", vowels);
console.log("Consonants:", consonants);
console.log("Digits:", digits);
console.log("Special characters:", special);
```

```python
Q2. Find the most frequent character

Given:

let str = "programming";
Find the character that occurs most frequently.

Expected:

Most frequent character: r
Frequency: 2
 
```

```python
let str = "programming";

let frequency = {};
let maxChar = "";
let maxCount = 0;

for (let char of str) {
    frequency[char] = (frequency[char] || 0) + 1;

    if (frequency[char] > maxCount) {
        maxCount = frequency[char];
        maxChar = char;
    }
}

console.log("Most frequent character:", maxChar);
console.log("Frequency:", maxCount);
```

```python
Q4. Check whether two strings are anagrams

Two strings are anagrams if they contain the same characters with the same frequency.

Example:

"listen"
"silent"
```

```python
let str1 = "listen";
let str2 = "silent";

str1 = str1.toLowerCase();
str2 = str2.toLowerCase();

let sorted1 = str1.split("").sort().join("");
let sorted2 = str2.split("").sort().join("");

if (sorted1 === sorted2) {
    console.log("The strings are anagrams");
} else {
    console.log("The strings are not anagrams");
}
```
