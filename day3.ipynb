```python
#find the second largest element in a tuple without converting it to a list or using sort 

numbers = (10,20,30,28,23,44,34)

largest = second = float('-inf')

for num in numbers:
    if num > largest:
        second = largest
        largest = num
    elif (num>second) and (num!=largest):
        second = num

print ("second largest :",second)
```

```python
#remove duplicates from tuple without changing order of tuple

original_tuple = (1, 3, 5, 2, 3, 5, 1, 1, 3)

seen = set()
unique_elements = []

for elem in original_tuple:
    if elem not in seen:
        seen.add(elem)
        unique_elements.append(elem)

res = tuple(unique_elements)
print(res) 
```

```python
#find all pairs of numbers in a tuple whose sum is equal to a target element
def find_pairs_set(lst, target):
    seen = set()
    pairs_found = set()
    for num in lst:
        complement = target - num
        if complement in seen:
            # Normalize pair to (min, max) to ensure uniqueness
            pairs_found.add((min(num, complement), max(num, complement)))
        seen.add(num)
    return list(pairs_found)

# Example
nums = [1, 5, 7, -1, 5]
print(find_pairs_set(nums, 6)) 
```
